### 18. Calcula este promedio con Numpy y después con un método de Pandas

```python
import numpy as np

# Promedio con Numpy
promedio_numpy = np.mean(df['Rating'].dropna())
print("Promedio con Numpy:", promedio_numpy)

# Promedio con Pandas
promedio_pandas = df['Rating'].mean()
print("Promedio con Pandas:", promedio_pandas)
```

### 21. Obten los valores únicos de la variable Rating y la frecuencia total de cada uno de estos valores. Posteriormente crea un nuevo dataframe con essos valores

```python
# Obtener la frecuencia de cada rating
frecuencias = df['Rating'].value_counts().sort_index()

# Crear el nuevo DataFrame
df_frecuencias = pd.DataFrame({'Rating': frecuencias.index, 'Frecuencia': frecuencias.values})

print(df_frecuencias)
```

### 22. Ordena el nuevo dataframe por la variable `rating`

```python
df_frecuencias = df_frecuencias.sort_values('Rating')
print(df_frecuencias)
```

### 23. Crea una gráfica de barras con este nuevo dataframe ordenado

```python
plt.bar(df_frecuencias['Rating'], df_frecuencias['Frecuencia'])
plt.xlabel('Rating')
plt.ylabel('Frecuencia')
plt.title('Frecuencia de Ratings')
plt.show()
```

### 24. Crea la matriz de correlación del dataframe de películas

```python
# Crear la matriz de correlación del dataframe de películas (solo columnas numéricas)
matriz_correlacion = df.select_dtypes(include=['number']).corr()
print(matriz_correlacion)
```

### 25. Grafica la matriz de correlación utilizando `matshow()`

```python
# Graficar la matriz de correlación utilizando matshow()
import matplotlib.pyplot as plt
plt.matshow(matriz_correlacion)
plt.colorbar()
plt.xticks(range(len(matriz_correlacion.columns)), matriz_correlacion.columns, rotation=90)
plt.yticks(range(len(matriz_correlacion.index)), matriz_correlacion.index)
plt.title('Matriz de correlación')
plt.show()
```
